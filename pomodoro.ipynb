```python
import tkinter as tk

WORK_MIN = 25      # minutes for focus
BREAK_MIN = 5      # minutes for short break

class PomodoroApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Pomodoro Timer")

        # State
        self.is_running = False
        self.is_work = True          # True=Work, False=Break
        self.remaining = WORK_MIN * 60
        self.timer_id = None

        # UI
        self.mode_label = tk.Label(root, text="Work", font=("Arial", 18, "bold"))
        self.mode_label.pack(pady=(12, 0))

        self.time_label = tk.Label(root, text=self.format_time(self.remaining),
                                   font=("Arial", 48), fg="tomato")
        self.time_label.pack(pady=8)

        btn_frame = tk.Frame(root)
        btn_frame.pack(pady=6)

        self.start_btn = tk.Button(btn_frame, text="Start", width=8, command=self.start)
        self.pause_btn = tk.Button(btn_frame, text="Pause", width=8, command=self.pause)
        self.reset_btn = tk.Button(btn_frame, text="Reset", width=8, command=self.reset)

        self.start_btn.grid(row=0, column=0, padx=5)
        self.pause_btn.grid(row=0, column=1, padx=5)
        self.reset_btn.grid(row=0, column=2, padx=5)

        self.status_label = tk.Label(root, text="Ready", font=("Arial", 10))
        self.status_label.pack(pady=(6, 12))

        self.update_display()

    def format_time(self, secs):
        m = secs // 60
        s = secs % 60
        return f"{m:02d}:{s:02d}"

    def start(self):
        if not self.is_running:
            self.is_running = True
            self.status_label.config(text="Running")
            self.tick()

    def pause(self):
        if self.is_running:
            self.is_running = False
            self.status_label.config(text="Paused")
            if self.timer_id:
                self.root.after_cancel(self.timer_id)
                self.timer_id = None

    def reset(self):
        self.pause()
        self.is_work = True
        self.remaining = WORK_MIN * 60
        self.status_label.config(text="Ready")
        self.update_display()

    def switch_mode(self):
        self.is_work = not self.is_work
        self.remaining = (WORK_MIN if self.is_work else BREAK_MIN) * 60
        self.mode_label.config(text="Work" if self.is_work else "Break")
        self.time_label.config(fg="tomato" if self.is_work else "green")

    def tick(self):
        if not self.is_running:
            return
        if self.remaining > 0:
            self.remaining -= 1
            self.update_display()
            self.timer_id = self.root.after(1000, self.tick)
        else:
            # Auto-switch when timer hits zero
            self.switch_mode()
            self.update_display()
            self.timer_id = self.root.after(1000, self.tick)

    def update_display(self):
        self.mode_label.config(text="Work" if self.is_work else "Break")
        self.time_label.config(text=self.format_time(self.remaining))

# Launch the app window
root = tk.Tk()
app = PomodoroApp(root)
root.mainloop()
```
